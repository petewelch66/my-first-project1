```python
#题目一：四位自然数处理
num = int(input("请输入一个四位自然数："))  

ge = num % 10          
shi = num // 10 % 10   
bai = num // 100 % 10  
qian = num // 1000     

new_num = ge * 1000 + shi * 100 + bai * 10 +qian

print("原数是：", num)
print("反转后的新数是：", new_num)

j = qian *shi
print("偶数位数字之积是：", j)
```

```python
#题目二：圆的周长和面积计算
import math  

r = float(input("请输入圆的半径："))

zhouchang = 2 * math.pi * r  
mianji = math.pi * r ** 2      

print(f"圆的周长为：{zhouchang:.3f}")
print(f"圆的面积为：{mianji:.3f}")
```

```python
#题目三：中国手机号格式判断
phone = input("请输入手机号：")

if not phone.isdigit():
    print("手机号格式不合法：不能包含字母、符号或空格！")
elif len(phone) != 11:
    print("手机号格式不合法：必须是11位！")
elif phone[0] != "1":
    print("手机号格式不合法：第一位必须是1！")
elif phone[1] not in ["3", "4", "5", "7", "8"]:
    print("手机号格式不合法：第二位只能是3、4、5、7、8！")
else:
    print("手机号格式合法！")
```
