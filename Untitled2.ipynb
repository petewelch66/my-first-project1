```python
def is_prime(n):
    if n<2:
        return False
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return False
    return True

start = int(input("请输入区间起始值："))
end = int(input("请输入区间结束值："))

primes = []
for num in range(start,end+1):
    if is_prime(num):
        primes.append(num)

print(f"区间[{start}, {end}]内共有{len(primes)}个素数，分别是：{primes}")
```

```python
n=int(input("请输入要分解的整数："))
original=n
factors=[]

i=2
while i*i<=n:
    while n%i==0:
        factors.append(str(i))
        n=n//i
    i+=1
if n>1:
    factors.append(str(n))

print(f"{original}={'*'.join(factors)}")
```

```python
X = [[12, 7, 3],
     [4, 5, 6],
     [7, 8, 9]]

Y = [[5, 8, 1],
     [6, 7, 3],
     [4, 5, 9]]

result=[]
for i in range(len(X)):
    row=[]
    for j in range(len(X[i])):
        row.append(X[i][j] + Y[i][j])
    result.append(row)

print("矩阵相加结果：")
for row in result:
    print(row)
```

```python
import random

num_list=random.sample(range(1, 101),20)
print("原始列表：",num_list)

odds=sorted([x for x in num_list if x%2!=0])
evens=sorted([x for x in num_list if x%2==0])

result=odds+evens
print("调整后列表：",result)
```

```python
vocabulary={}  

def show_menu():
    print("\n===== 生词本菜单 =====")
    print("1. 查看生词列表")
    print("2. 背单词")
    print("3. 添加新单词")
    print("4. 删除单词")
    print("5. 清空生词本")
    print("0. 退出程序")
    return input("请输入功能编号：")

def show_vocab():
    if not vocabulary:
        print("生词本为空！")
        return
    print("\n生词列表：")
    for idx, (en, cn) in enumerate(vocabulary.items(), 1):
        print(f"{idx}.{en}-{cn}")

def recite_words():
    if not vocabulary:
        print("生词本为空！")
        return
    print("\n开始背单词：")
    for en, cn in vocabulary.items():
        input(f"英文:{en}，按回车查看中文...")
        print(f"中文：{cn}\n")

def add_word():
    en = input("请输入英文单词：")
    if en in vocabulary:
        print(f"单词 {en} 已存在！")
        return
    cn = input("请输入中文释义：")
    vocabulary[en]=cn
    print(f"添加成功：{en}-{cn}")

def delete_word():
    en = input("请输入要删除的英文单词：")
    if en in vocabulary:
        del vocabulary[en]
        print(f"删除成功：{en}")
    else:
        print(f"单词 {en} 不存在！")

def clear_vocab():
    confirm = input("确认清空生词本？(y/n)：")
    if confirm.lower() == "y":
        vocabulary.clear()
        print("生词本已清空！")
    else:
        print("已取消清空操作")

while True:
    choice=show_menu()
    if choice=="1":
        show_vocab()
    elif choice=="2":
        recite_words()
    elif choice=="3":
        add_word()
    elif choice=="4":
        delete_word()
    elif choice=="5":
        clear_vocab()
    elif choice=="0":
        print("退出程序，再见！")
        break
    else:
        print("输入错误，请重新选择！")
```
