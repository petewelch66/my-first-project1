```python
#示例一：判断学生成绩等级
score=int(input("请输入学生的成绩："))
if score<0 or score>100:
    print("成绩有误！")
elif 0<=score<60:
    print("E")
elif 60<=score<70:
    print("D")
elif 70<=score<80:
    print("C")
elif 80<=score<90:
    print("B")
else:
    print("A")
```

```python
#示例二：判断回文数
number=input("Please input a number.")
num_reverse=number[::-1]
if number==num_reverse:
    print("{number}是回文数")
else:
    print("{}不是回文数".format(number))
```

```python
#实例三：BMI值计算
weight = float(input("请输入你的体重（单位：kg）:"))
height = float(input("请输入你的身高（单位：m）:"))
BMI = weight / (height**2)
if BMI < 18.5:
    print(f'您的BMI指数是{BMI:.2f}，表示体重过轻！')
elif 18.5 <= BMI <= 23.9:
    print(f'您的BMI指数是{BMI:.2f}，表示体重正常！')
elif 24 <= BMI <= 27.9: 
    print(f'您的BMI指数是{BMI:.2f}，表示体重过重！')
elif 28 <= BMI <= 32:
    print(f'您的BMI指数是{BMI:.2f}，表示肥胖！')
else:
    print(f'您的BMI指数是{BMI:.2f}，表示非常肥胖！')
```

```python
#实例四：模拟乘客进站
ticket = int(input("请出示您的车票（1代表有车票，0代表无车票）:"))
if ticket == 1:
    print("验票通过")
    safe = input("您是否携带危险物（y,Y/n,N）")
    if safe == 'y' or safe == 'Y':
        print("携带危险物，请等待处理！")
    elif safe == 'n' or safe == 'N':
        print("通过安检，请去候车厅候车")
    else:
        print("输入错误")
else:
    print("请先买票！")
```

```python
#实例五：快递计费
weight = eval(input("请输入快递的重量（kg）:"))
sum = 0
ID = input("请输入寄件区域编号(01华东地区,02华南地区,03华北地区)：")
if ID == "01":
    if weight <= 2:
        money = sum+13
    else:
        money = (int(13+(weight-2)*3))
        print("请支付快递费%d" % money+"元！")
elif ID == "02":
    if weight <= 2:
        money = sum+12
    else:
        money = (int(12+(weight-2)*2))
        print("请支付快递费{money}元！")
elif ID == "03":
    if weight <= 2:
        money = sum + 14
    else:
        money = (int(14 + (weight - 2) * 4))
        print("请支付快递费{}" .format(money)+ "元！")
else:
    print("地区编号错误")
```

```python
#实例六：数据加密
pwd = input("请输入密码：")
sum = 0
a = ''
for i in pwd:
    sum += ord(i) 
    a += str(ord(i)) 
a = a[::-1] 
a = int(a)
pwd = a+sum 
print(f"加密后的密码为：{pwd}")
```

```python
#实例七：逢三或七拍手
for num in range(1, 101):
    if '3' in str(num) or '7' in str(num) or num % 3 == 0 or num % 7 == 0:
        print("**", end="\t")
    else:
        print(num, end='\t')
    if num % 10 == 0: 
        print(" ")
```

```python
#实例八：登录账号检测
i = 0 
while i < 3: 
    user_name = input("请输入用户名：")
    user_pwd = input("请输入密码：")
    if user_name == 'trumputin' and user_pwd == '12345':
        print("系统登录中，请稍后...")
        break
    elif user_name != 'trumputin':
        print("用户名不存在，您还有",2-i,"次机会，请重新输入！")
    elif user_name == 'trumputin' and user_pwd != '12345':
        print("密码错误，您还有",2-i,"次机会，请重新输入！")
    i += 1 
else:
    print("对不起，3次机会均输入错误，请24小时后再试！")
```

```python
#实例十：猜数游戏
import random
rand = random.randint(1,50)
count = 1
while count <=5:
    number = eval(input('请猜测随机生成的数字：'))
    if number== rand:
        print('猜对了')
        break
    elif number < rand:
        print('小了')
    elif number > rand:
        print('大了')
    count+=1
if count<=2:
    print('真聪明，一共猜了',count,'次')
elif count<=4:
    print('还可以，一共猜了',count,'次')
else:
    print('次数太多了，一共用了',count,'次')
```
