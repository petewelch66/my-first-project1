```python
def peven(n):
    s=0.0
    for i in range(1,n+1,2):
        s+=1.0/i
    return s
    
def podd(n):
    s=0.0
    for i in range(1,n+1,2):
        s+=1.0/i
    return s
        
n=int(input("请输入一个数:"))
if n%2==0:
    print(peven(n))
else:
    print(podd(n))
```

```python
import math
def func(a,b,c):
    deta=b*b-4*a*c
    if deta<0:
        return "方程无解"
    elif deta==0:
        return (-b/2*a)
    else:
        detaa=math.sqrt(deta)
        x1=(-b/2*a)+(detaa/2*a)
        x2=(-b/2*a)+(-detaa/2*a)
        return x1,x2

a=int(input("请输入a:"))
b=int(input("请输入b:"))
c=int(input("请输入c:"))

result=func(a,b,c)
print("方程的结果为：",result)
```
